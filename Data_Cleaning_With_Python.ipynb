```python
df = df.drop_duplicates()
```

```python
df = df.dropna(subset=['CustomerID'])
```

```python
df['CustomerID'] = df['CustomerID'].astype(int)
```

```python
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]
```

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']
```

```python
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)].copy()
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']
```

```python
len(df)
```

```python
import re
```

```python
df['Description'] = df['Description'].apply(lambda x: re.sub(r'[^a-zA-Z0-9\s]', '', str(x)).strip() if pd.notna(x) else x)
df['StockCode'] = df['StockCode'].astype(str).str.replace(r'[^0-9]', '', regex=True)
df = df[df['StockCode'] != ''].copy()
```

```python
df.head()
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Clean_Data'] = df['InvoiceDate'].dt.date
df['Purchase_Hour'] = df['InvoiceDate'].dt.hour
df = df.drop(columns=['InvoiceDate']).copy()
```
